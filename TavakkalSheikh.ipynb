```python
# TASK Title: Data Cleaning & Preprocessing
# Dataset: Titanic Dataset
import pandas as pd
df = pd.read_csv(r"C:\Users\DELL\Desktop\Titanic-Dataset.csv")
print("First 5 Rows of Dataset:")
print(df.head())
print("\nDataset Information:")
print(df.info())
```

```python
# TASK 1: HANDLE MISSING VALUES
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df["Cabin"] = df["Cabin"].fillna("Unknown")
print("\nMissing values handled successfully!")
```

```python
# TASK 2: REMOVE DUPLICATES
before = len(df)
df = df.drop_duplicates()
after = len(df)
print("\nDuplicate rows removed:", before - after)
```

```python
# TASK 3: CONVERT DATA TYPES
df["Survived"] = df["Survived"].astype(str)
print("\nData type of Survived column changed to String.")
```

```python
# TASK 4: RENAME COLUMNS
df = df.rename(columns={
    "PassengerId": "Passenger_ID",
    "Pclass": "Passenger_Class",
    "Sex": "Gender"
})
print("\nColumns renamed successfully!")
```

```python
# CHECK CLEANED DATASET
print("\nCleaned Dataset:")
print(df.head())
print("\nCleaned Dataset Information:")
print(df.info())
```

```python
# SAVE CLEANED DATASET
df.to_csv("Cleaned_Titanic_Dataset.csv", index=False)
print("\nData Cleaning and Preprocessing Completed Successfully!")
print("Cleaned dataset saved as 'Cleaned_Titanic_Dataset.csv'")
```
